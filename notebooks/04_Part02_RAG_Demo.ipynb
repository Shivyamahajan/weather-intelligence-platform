```python
from langchain_ollama import OllamaLLM

llm = OllamaLLM(
    model="qwen2.5:1.5b",
    temperature=0.1
)

question = "What is the IMD classification for 85mm rainfall?"

print(f"Question: {question}\n")
print("LLM Answer WITHOUT RAG:")
print(llm.invoke(question))
```

```python
from src.genai.rag_pipeline import WeatherRAGPipeline

rag = WeatherRAGPipeline()

response = rag.ask(question)

print(f"Question: {question}\n")
print("RAG Answer (grounded in documents):")
print(response.answer)

print(f"\nSources used: {response.sources}")

print(
    f"\nTime breakdown: "
    f"Retrieval={response.retrieval_time_ms:.0f}ms, "
    f"Generation={response.generation_time_ms:.0f}ms"
)
```

```python
comparison_questions = [
    "What is the IMD rainfall warning colour system?",
    "When does monsoon arrive in Chennai?",
    "Which ML model was best for this project?",
]

for q in comparison_questions:
    print(f"\n{'=' * 60}")
    print(f"Question: {q}")
    print(f"{'=' * 60}")

    no_rag = llm.invoke(q)
    rag_resp = rag.ask(q)

    print(f"\nWITHOUT RAG:\n{no_rag[:300]}...")
    print(f"\nWITH RAG:\n{rag_resp.answer[:300]}...")
    print(f"\nRAG is grounded in: {rag_resp.sources}")
```

```python
from src.genai.conversational_assistant import ConversationalWeatherAssistant

assistant = ConversationalWeatherAssistant(memory_window=3)

turns = [
    "What is very heavy rainfall according to IMD?",
    "What damage can it cause?",
    "Which Indian cities are most vulnerable to this?",
]

for turn in turns:
    print(f"\n👤 {turn}")
    response = assistant.chat(turn)
    print(f"🤖 {response[:400]}...")
```

```python
prediction_data = {
    'city': 'Mumbai',
    'date': '2026-08-07',
    'predicted_mm': 145.0,
    'category': 'Heavy Rain'
}

question = (
    "Given this prediction, what actions should city authorities take?"
)

response = assistant.chat(question, prediction_data)

print(f"Question: {question}")
print(f"\nPrediction provided to AI: {prediction_data}")
print(f"\nAI Response:")
print(response)
```
